```python
from pathlib import Path
import pandas as pd
import numpy as np
from PIL import Image
from tqdm.auto import tqdm

# ------------------------------------------------------------------
# Paths
# ------------------------------------------------------------------

CSV_PATH = Path(r"C:\Users\user\Desktop\LUCY\AI Unleashed 2026\datasets\fer2013_clean.csv")
OUTPUT_DIR = Path(r"C:\Users\user\Desktop\LUCY\AI Unleashed 2026\datasets\FER2013_CLEAN")

# ------------------------------------------------------------------
# Emotion mapping
# ------------------------------------------------------------------

emotion_names = {
    0: "angry",
    1: "disgust",
    2: "fear",
    3: "happy",
    4: "sad",
    5: "surprise",
    6: "neutral"
}

# ------------------------------------------------------------------
# Load CSV
# ------------------------------------------------------------------

df = pd.read_csv(CSV_PATH)

# ------------------------------------------------------------------
# Create folders
# ------------------------------------------------------------------

for emotion in emotion_names.values():
    (OUTPUT_DIR / emotion).mkdir(parents=True, exist_ok=True)

print("Folders created.")

# ------------------------------------------------------------------
# Convert every row into PNG
# ------------------------------------------------------------------

for idx, row in tqdm(df.iterrows(), total=len(df)):

    pixels = np.fromstring(
        row["pixels"],
        sep=" ",
        dtype=np.uint8
    ).reshape(48, 48)

    img = Image.fromarray(pixels)

    emotion = emotion_names[row["emotion"]]

    filename = f"row_{idx:05d}.png"

    img.save(
        OUTPUT_DIR / emotion / filename,
        compress_level=0   # lossless PNG
    )

print("\nDataset generation completed successfully.")
print(f"Saved to:\n{OUTPUT_DIR}")
```
